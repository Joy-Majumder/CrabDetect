```python
from pathlib import Path
from collections import Counter

labels_dir = NEW_DATASET / "train" / "labels"

format_counts = Counter()
polygon_examples = []
detection_examples = []

for label_file in labels_dir.glob("*.txt"):
    for line in label_file.read_text().splitlines():
        values = line.split()

        if not values:
            continue

        n = len(values)

        if n == 5:
            format_counts["detection"] += 1
            if len(detection_examples) < 3:
                detection_examples.append(label_file.name)

        elif n > 5 and (n - 1) % 2 == 0:
            format_counts["segmentation"] += 1
            if len(polygon_examples) < 3:
                polygon_examples.append(label_file.name)

        else:
            format_counts["other"] += 1

print("Annotation format counts:")
for k, v in format_counts.items():
    print(f"{k}: {v}")

print("\nDetection examples:")
for x in detection_examples:
    print(x)

print("\nSegmentation examples:")
for x in polygon_examples:
    print(x)
```

```python
from pathlib import Path
from collections import Counter

DATASET = Path("/kaggle/input/datasets/joy0x1/crabfinaldataset/Crabs Final Test.v1i.yolo26")

for split in ["train", "valid", "test"]:
    labels_dir = DATASET / split / "labels"

    counts = Counter()

    for label_file in labels_dir.glob("*.txt"):
        for line in label_file.read_text().splitlines():
            values = line.split()

            if not values:
                continue

            if len(values) == 5:
                counts["detection"] += 1
            elif len(values) > 5 and (len(values) - 1) % 2 == 0:
                counts["segmentation"] += 1
            else:
                counts["other"] += 1

    print(f"\n{split.upper()}")
    print("-" * 30)
    print(f"Detection:    {counts['detection']}")
    print(f"Segmentation: {counts['segmentation']}")
    print(f"Other:        {counts['other']}")
```

```python
from pathlib import Path
import shutil

SOURCE = Path(
    "/kaggle/input/datasets/joy0x1/crabfinaldataset/"
    "Crabs Final Test.v1i.yolo26"
)

OUTPUT = Path("/kaggle/working/crab_final_detection")

CLASS_MAP = {
    0: 0,  # Green -> Green
    1: 2,  # Jonah -> Jonah
    2: 1,  # Rock -> Rock
}

for split in ["train", "valid", "test"]:
    (OUTPUT / split / "images").mkdir(parents=True, exist_ok=True)
    (OUTPUT / split / "labels").mkdir(parents=True, exist_ok=True)


def polygon_to_bbox(values):
  

    cls = int(values[0])
    coords = list(map(float, values[1:]))

    xs = coords[0::2]
    ys = coords[1::2]

    xmin = min(xs)
    xmax = max(xs)
    ymin = min(ys)
    ymax = max(ys)

    x_center = (xmin + xmax) / 2
    y_center = (ymin + ymax) / 2
    width = xmax - xmin
    height = ymax - ymin

    return cls, x_center, y_center, width, height


stats = {}

for split in ["train", "valid", "test"]:

    src_images = SOURCE / split / "images"
    src_labels = SOURCE / split / "labels"

    out_images = OUTPUT / split / "images"
    out_labels = OUTPUT / split / "labels"

    detection_count = 0
    segmentation_count = 0
    image_count = 0
    label_count = 0

    for image_file in src_images.iterdir():

        if not image_file.is_file():
            continue

        shutil.copy2(
            image_file,
            out_images / image_file.name
        )

        image_count += 1

    for label_file in src_labels.glob("*.txt"):

        output_lines = []

        for line in label_file.read_text().splitlines():

            if not line.strip():
                continue

            values = line.split()

            if len(values) == 5:

                old_cls = int(values[0])

                xc = float(values[1])
                yc = float(values[2])
                w = float(values[3])
                h = float(values[4])

                detection_count += 1

            else:

                old_cls, xc, yc, w, h = polygon_to_bbox(values)

                segmentation_count += 1

            new_cls = CLASS_MAP[old_cls]

            output_lines.append(
                f"{new_cls} "
                f"{xc:.6f} "
                f"{yc:.6f} "
                f"{w:.6f} "
                f"{h:.6f}"
            )

            label_count += 1

        (out_labels / label_file.name).write_text(
            "\n".join(output_lines) + "\n"
        )

    stats[split] = {
        "images": image_count,
        "labels": label_count,
        "detection": detection_count,
        "segmentation_converted": segmentation_count,
    }


print("Dataset conversion complete!\n")

for split, s in stats.items():

    print(split.upper())
    print("-" * 40)
    print("Images:                  ", s["images"])
    print("Objects:                 ", s["labels"])
    print("Original detection:      ", s["detection"])
    print("Polygon → bbox converted:", s["segmentation_converted"])
    print()
```
